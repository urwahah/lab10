# Estimating π by Throwing Darts

Picture a square target with a circle inscribed inside it, and imagine
throwing darts at it completely at random. Some land inside the circle,
some land in the corners outside it. The *fraction* that land inside the
circle turns out to be enough to estimate π. We don't require geometry formulas,
just random points and some counting.


```{tip} Reproducibility note
:class: dropdown
This simulation uses *randomness*, which means it needs a fixed seed to be
reproducible. Without `np.random.seed(...)`, rerunning this cell would give
a slightly different estimate of π every time, and no one could reproduce today's exact number.
```


````python
import numpy as np

np.random.seed(159)   # fixed seed: same "random" points every run
n_darts = 10_000

x = np.random.uniform(-1, 1, n_darts)
y = np.random.uniform(-1, 1, n_darts)
inside_circle = (x**2 + y**2) <= 1

pi_estimate = 4 * inside_circle.sum() / n_darts
print(f"Darts thrown: {n_darts:,}")
print(f"Estimated pi: {pi_estimate:.5f}")
````

The true value of π is 3.14159.... With 10,000 random points, the
estimate above usually lands within about 1% of that.

This technique is called a **Monte Carlo method** — using repeated random
sampling to estimate something that's hard to compute directly. It shows up
everywhere in statistics, from estimating integrals to simulating confidence
intervals.
